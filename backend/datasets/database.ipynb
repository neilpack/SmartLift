```python
import sqlite3
import pandas as pd

# Load your CSV
df = pd.read_csv("exercises_200.csv")

# Connect to SQLite (this creates exercises.db if it doesn’t exist)
conn = sqlite3.connect("exercises.db")
c = conn.cursor()

# Create the exercises table
c.execute("""
CREATE TABLE IF NOT EXISTS exercises (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    exercise_name TEXT,
    equipment TEXT,
    muscle_group TEXT,
    description TEXT
)
""")

# Insert data into the table
df.to_sql("exercises", conn, if_exists="replace", index=False)

conn.commit()
conn.close()

print("✅ Database created: exercises.db")
```

```python
conn = sqlite3.connect("exercises.db")
c = conn.cursor()

# Example: Get all dumbbell chest exercises
c.execute("SELECT * From exercises")
rows = c.fetchall()

for row in rows:
    print(row)

conn.close()
```
